```python
import math

# Robot arm lengths
L1 = 10
L2 = 8

# Joint angles in degrees
theta1 = 30
theta2 = 45

# Convert degrees to radians
t1 = math.radians(theta1)
t2 = math.radians(theta2)

# Calculate X and Y position
x = L1 * math.cos(t1) + L2 * math.cos(t1 + t2)
y = L1 * math.sin(t1) + L2 * math.sin(t1 + t2)

# Display result
print("===== ROBOT ARM POSITION CALCULATOR =====")
print("First Arm Length :", L1)
print("Second Arm Length:", L2)
print("First Angle      :", theta1)
print("Second Angle     :", theta2)

print("\n===== RESULT =====")
print("X Position =", round(x, 2))
print("Y Position =", round(y, 2))
```
